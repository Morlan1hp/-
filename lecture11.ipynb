# Merge Sort

```python
def Merge(B, C, A):
    # A is an array, B is an array, C is an array
    p = len(B)
    q = len(C)
    i = j = k = 0     # k indexes A, i indexes B, j indexes C
    while i < p and j < q:
        snapshot()
        if B[i] <= C[j]:
            A[k] = B[i]
            i = i + 1
        else:
            A[k] = C[j]
            j = j + 1
        k = k + 1
    snapshot()
    if i == p:
        A[k: p + q] = C[j: q]
        snapshot()
    else:
        A[k: p + q] = B[i: p]
        snapshot()


def Mergesort(A):
    if len(A) > 1:
        n = len(A)
        B = A[:n // 2]
        C = A[n // 2:]
        snapshot(printstack=True)
        Mergesort(B)
        snapshot(printstack=True)
        Mergesort(C)
        snapshot(printstack=True)
        Merge(B, C, A)
        snapshot(printstack=True)

clear_frames()

array = [3, 1, 4, 1, 5, 9, 2, 6]
Mergesort(array)

show_animation(size=[8,10])
```

# Hoare Partition

```python
def HoarePartition(A, lo, hi):
    # A is an array
    p = A[lo]
    i = lo
    j = hi
    # i indexes A
    # j indexes A
    while i < j:
        snapshot()
        while i < hi and A[i] <= p:
            i = i + 1
            snapshot()
        while j >= lo and A[j] > p:
            j = j - 1
            snapshot()
        A[i], A[j] = A[j], A[i]
        snapshot()
    A[i], A[j] = A[j], A[i]
    snapshot()
    A[lo], A[j] = A[j], A[lo]
    snapshot()
    return j


clear_frames()
array = [3, 1, 4, 5, 9, 2, 6, 8]
HoarePartition(array, 0, len(array) - 1)
show_animation(size=[8,7])
```

# Quick Sort

```python

# Note we could also use LomutoPartition from lecture 10 but
# using Hoare's partitioning algorithm is more traditional
def Quicksort(A, lo, hi):
    # A is an array
    snapshot(printstack=True)
    if lo < hi:
        s = HoarePartition(A, lo, hi)
        Quicksort(A, lo, s - 1)
        Quicksort(A, s + 1, hi)


clear_frames()

array = [3, 1, 4, 5, 9, 2, 6, 8]
Quicksort(array, 0, len(array) - 1)

show_animation(size=[12,9])
```
