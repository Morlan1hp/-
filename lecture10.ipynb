# Binary Search

```python
def BinSearch(A, k):
    # A is an array
    lo = 0
    hi = len(A) - 1
    # lo indexes A, hi indexes A
    snapshot()
    while lo <= hi:
        # m indexes A
        m = lo + (hi - lo) // 2
        snapshot()
        if A[m] == k:
            snapshot()
            return m
        if A[m] > k:
            hi = m - 1
            snapshot()
        else:
            lo = m + 1
            snapshot()
    return -1

clear_frames()
BinSearch([1, 2, 3, 5, 8, 13], 8)
show_animation(size=[8,7])
```

# Lomuto Partition

```python
def LomutoPartition(A, lo, hi):
    # A is an array, lo indexes A, hi indexes A
    p = A[lo]
    s = lo # s indexes A
    snapshot()
    # i indexes A
    for i in range(lo + 1, hi + 1):
        snapshot()
        if A[i] < p:
            s = s + 1
            snapshot()
            A[s], A[i] = A[i], A[s]
            snapshot()
    A[s], A[lo] = A[lo], A[s]
    snapshot()
    return s

clear_frames()

array = [3, 1, 4, 5, 9, 2, 6, 8]
lo = 0
hi = len(array) - 1

LomutoPartition(array, lo, hi)

show_animation(size=[8,6])
```

# Quick Select

```python
def QuickSelect(A, lo, hi, k):
    # A is an array, lo indexes A, hi indexes A
    snapshot(printstack=True)
    # s indexes A
    s = LomutoPartition(A, lo, hi)
    if s - lo == k - 1:
        snapshot(printstack=True)
        return A[s]
    else:
        if s - lo > k - 1:
            snapshot(printstack=True)
            return QuickSelect(A, lo, s - 1, k)
        else:
            snapshot(printstack=True)
            return QuickSelect(A, s + 1, hi, (k - 1) - (s - lo))

clear_frames()
array = [3, 1, 4, 5, 9, 2, 6, 8]
QuickSelect(array, 0, len(array) - 1, 7)
show_animation(size=[12,7])
```

# Interpolation Search

```python
def InterpolationSearch(A, k):
    # A is an array
    # lo indexes A
    # hi indexes A
    lo = 0
    hi = len(A) - 1
    while lo < hi:
        snapshot()
        if A[lo] > k or A[hi] < k:
            break
        if lo == hi:
            if A[lo] == k:
                snapshot()
                return lo
            else:
                break
        m = lo + (k - A[lo]) * (hi - lo) // (A[hi] - A[lo]) # m indexes A
        snapshot()
        if A[m] == k:
            snapshot()
            return m
        if A[m] > k:
            hi = m - 1
            snapshot()
        else:
            lo = m + 1
            snapshot()
    return -1

clear_frames()
for i in range(15):
    InterpolationSearch([1, 2, 3, 5, 8, 13], i)
show_animation(size=[8,10])
```
