# Brute-Force String Matching

```python
def StringMatching(pattern, text):
    # pattern is a character array
    # text is a character array
    m = len(pattern)
    n = len(text)
    for i in range(0, n - m + 1): # i indexes text
        j = 0
        snapshot()
        while j < m and pattern[j] == text[i + j]:  # j indexes pattern
            j = j + 1
            snapshot()
        if j == m:
            snapshot()
            return i
    snapshot()
    return -1

clear_frames()
StringMatching("hello", "abchellodef")
show_animation(size=[10,7])
```

```python
def ClosestPair(nodes):
    min = float("inf")
    n = len(nodes)
    for i in range(0, n - 1):
        xi, yi = nodes[i]
        for j in range(i + 1, n):
            xj, yj = nodes[j]
            d = ((xi - xj) ** 2 + (yi - yj) ** 2) ** 0.5
            snapshot()
            if d < min:
                min = d
                pi = i
                pj = j
    snapshot()
    return pi, pj

clear_frames()
N = [(1, 3), (5, 8), (-2, 7), (-9, 0), (-4, -6)]
ClosestPair(N)
show_animation(size=[18,5])
```
